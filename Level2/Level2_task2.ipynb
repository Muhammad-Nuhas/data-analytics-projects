```python
import pandas as pd
df=pd.read_csv('D:/vs code1/internship/Dataset .csv')
df
```

```python
cuisine_combinations=df['Cuisines'].value_counts()
print("Most common cuisine combination:")
print(cuisine_combinations.head(10))
```

```python
avg_rating_by_cuisine=(df.groupby('Cuisines')['Aggregate rating'].mean().sort_values(ascending=False))
print("Cuisine combinations with highest average ratings:")
print(avg_rating_by_cuisine.head(10))
```
