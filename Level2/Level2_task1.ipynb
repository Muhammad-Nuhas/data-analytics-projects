```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('D:/vs code1/internship/Dataset .csv')
df
```

```python
plt.hist(df['Aggregate rating'],edgecolor='black')
plt.xlabel('Aggregate Rating')
plt.ylabel('Number of Restaurants')
plt.title('Distribution of Restaurant Ratings')
plt.show()
```

```python
bins = [0,1,2,3,4,5]
labels = ['0-1','1-2','2-3','3-4','4-5']
df['Rating Range']=pd.cut(df['Aggregate rating'],bins=bins,labels=labels)
most_common_range=df['Rating Range'].value_counts().index[0]
print("Most common rating range:",most_common_range)
```

```python
avg_votes=df['Votes'].mean()
print("Average number of votes:",avg_votes)
```
