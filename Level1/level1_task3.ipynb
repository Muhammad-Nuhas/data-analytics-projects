```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('D:/vs code1/internship/Dataset .csv')
df
```

```python

price_count=df['Price range'].value_counts().sort_index()
plt.bar(price_count.index,price_count.values)
plt.xlabel("Price range")
plt.ylabel("Number of restaurants")
plt.title("Distribution of price ranges")
plt.show()
```

```python
total=len(df)
price_percentage=(price_count/total)*100
print("Percentage of Restaurants in each price range:")
print(price_percentage)
```
