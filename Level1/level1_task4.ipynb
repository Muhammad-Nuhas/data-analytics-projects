```python
import pandas as pd
df=pd.read_csv('D:/vs code1/internship/Dataset .csv')
df
```

```python
online_counts=df['Has Online delivery'].value_counts()
online_counts
total=len(df)
percentage_online=(online_counts['Yes']/total)*100
percentage_offline=(online_counts['No']/total)*100
print("Percentage of online delivery:",percentage_online)
print("Percentage of offline delivery:",percentage_offline)
```

```python
avg_ratings=df.groupby('Has Online delivery')['Aggregate rating'].mean()
print("Average rating comparison:")
print(avg_ratings)
```
