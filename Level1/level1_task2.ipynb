```python
import pandas as pd
df=pd.read_csv('D:/vs code1/internship/Dataset .csv')
df
```

```python
top_city=df['City'].value_counts().index[0]
print("City with the most restaurants:",top_city)
```

```python
avg_rating_city=df.groupby('City')['Aggregate rating'].mean()
print("Average rating of each city:")
print(avg_rating_city)
```

```python
best_city=avg_rating_city.sort_values(ascending=False).index[0]
best_city_rating=avg_rating_city.max()
print("City with the highest average rating:", best_city)
print("Highest average rating:", best_city_rating)
```
